## Data Quality Checks

```python
# Check for unrealistic Year_Birth values (customers older than 100 years)

current_year = 2014  # last enrollment year in the dataset

customer_age = current_year - df["Year_Birth"]  # temporary variable, not added to df

invalid_age_count = (customer_age > 100).sum()

print(f"Customers older than 100 years: {invalid_age_count}")
print("\nOldest Year_Birth values:")
print(df["Year_Birth"].sort_values().head())
```

```python
# Check for invalid or extreme Income values

negative_income_count = (df["Income"] < 0).sum()

high_income_threshold = 600000
high_income_count = (df["Income"] > high_income_threshold).sum()

print(f"Negative Income values: {negative_income_count}")
print(f"Income values above {high_income_threshold}: {high_income_count}")
print(f"Maximum Income value: {df['Income'].max()}")
```

```python
# Check for inconsistent Marital_Status categories

marital_categories = df["Marital_Status"].unique()

print("Marital_Status categories found:")
print(marital_categories)

unusual_marital_values = ["Absurd", "YOLO", "Alone"]
unusual_marital_count = df["Marital_Status"].isin(unusual_marital_values).sum()

print(f"\nRows with unusual Marital_Status values: {unusual_marital_count}")
```

```python
# Create incorrect values report

incorrect_values_report = pd.DataFrame({
    "Check": [
        "Customers older than 100 years (Year_Birth)",
        "Negative Income values",
        "Income values above 600000",
        "Unusual Marital_Status categories (Absurd, YOLO, Alone)"
    ],
    "Count": [
        invalid_age_count,
        negative_income_count,
        high_income_count,
        unusual_marital_count
    ]
})

incorrect_values_report
```

```python
# Save incorrect values report

output_path = PROJECT_DIR / "03_Data_Quality" / "incorrect_values_report.xlsx"
output_path.parent.mkdir(parents=True, exist_ok=True)

incorrect_values_report.to_excel(output_path, index=False)

print("✅ Incorrect Values Report Saved!")
print(output_path)
```

## Numerical Variable Summary

```python
# Identify numerical columns

numerical_columns = df.select_dtypes(include=['number']).columns

print("Number of Numerical Variables:", len(numerical_columns))
print("\nNumerical Variables:\n")

for column in numerical_columns:
    print(column)
```

```python
# Generate summary statistics

numerical_summary = pd.DataFrame({
    "Variable": numerical_columns,
    "Minimum": df[numerical_columns].min().values,
    "Maximum": df[numerical_columns].max().values,
    "Mean": df[numerical_columns].mean().round(2).values
})

numerical_summary
```

```python
# Add simple observations

observations = []

for column in numerical_columns:
    minimum = df[column].min()
    maximum = df[column].max()

    observations.append(f"Values range from {minimum} to {maximum}")

numerical_summary["Observation"] = observations

numerical_summary
```

```python
# Save numerical summary

output_path = PROJECT_DIR / "02_Inspection_Report" / "numerical_variables_summary.xlsx"
output_path.parent.mkdir(parents=True, exist_ok=True)

numerical_summary.to_excel(output_path, index=False)

print("✅ Numerical Variables Summary Saved Successfully!")
print(output_path)
```
